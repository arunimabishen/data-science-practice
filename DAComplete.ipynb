```python
print('''
1: Find all the fctors of a given number
2: find common factors of two integers
3: list factors for all numbers from 1 to a given number''')
choice=int(input('Enter a choice form the given options : '))
def factors(n):
    list1=[]
    print('The factors are:')
    for i in range(1,n+1):
        if (n%i==0):
            list1.append(i)
    return list1

if(choice==1):
    n=int(input("Enter a number"))
    factors(n)

elif(choice==2):
    a=int(input("Enter first number"))
    b=int(input("Enter second number"))
    fact_a=factors(a)
    fact_b=factors(b)
    print('The common factors are:')
    for i in fact_a:
        if i in fact_b:
            print(i)
elif(choice==3):
    n=int(input("Enter a number"))
    dict={}
    for i in range(1,n+1):
        dict[i]=factors(i)
    print(dict)

else:
    print('No choice')
```

```python
#File Handling
try:
  f=open('myfile1.txt','w')
  print('Enter text(Type# when you are done)')

  s=""
  while s!='#':
    s=input()
    f.write(s+ '\n')

except FileNotFoundError:
  print('Error: The file was not found')
except IOError:
  print('Error: An IOError occured while handling the file.')
except Exception as e:
  print(f"Unexpected Error: {e}")
finally:
  print('\n File operation completed')
  f.close()
```

```python
"""Q.) Create a File, put some content in the file, copy the content of the first
file to another file display the content of the second file (This helps with the conept of deep and shallow copy)"""
# Create and open files
f1 = open('myfile_old.txt', 'w+')
f2 = open('myfile_new.txt', 'w+')

print("Enter text (Type # when you're done)")

while True:
    s = input()
    if s == '#':
        break
    f1.write(s + '\n')

# Copy content from first file to second
f1.seek(0)
content = f1.read()
f2.write(content)

# Display content of second file
f2.seek(0)
print("\nContent of second file:")
print(f2.read())

# Close files
f1.close()
f2.close()
```

```python
# The top _____ of the user are driving 20% of the value of transactions. This is equivalent to the value of transactions
# driven by the bottom _____ of the user. Hold true for both currencies (GBP & Euro).

user_sums = df.groupby('USER_ID')['new_amount'].sum().sort_values(ascending=False)

total_value = user_sums.sum()     #Total Transaction value

#For top 20
cumulative_sum = user_sums.cumsum()
top_users_count = (cumulative_sum <= 0.2 * total_value).sum() + 1
top_users_percent = (top_users_count / df['USER_ID'].nunique()) * 100

#for bottom 20
bottom_cumulative_sum = user_sums[::-1].cumsum()
bottom_users_count = (bottom_cumulative_sum <= 0.2 * total_value).sum() + 1
bottom_users_percent = (bottom_users_count / df['USER_ID'].nunique()) * 100

print(f"Top {top_users_percent:.2f}% of users drive 20% of transaction value.")
print(f"Bottom {bottom_users_percent:.2f}% of users drive 20% of transaction value.")
```

```python
# Q.Another data slice shows that ~50% of total transactions are driven by only ~13.5% of the total unique users.
user_totals = df.groupby('USER_ID')['new_amount'].sum().reset_index()
user_totals = user_totals.sort_values(by='new_amount', ascending=False)
user_totals['cumsum'] = user_totals['new_amount'].cumsum()
total_value = user_totals['new_amount'].sum()
user_totals['cumperc'] = user_totals['cumsum'] / total_value * 100
top_50_percent_users = user_totals[user_totals['cumperc'] <= 50]
fraction_users = len(top_50_percent_users) / df['USER_ID'].nunique() * 100
print(f"~{fraction_users:.2f}% of users drive ~50% of total transaction value")
```

## **LAB 11**

Perform the basic statistical analysis by counting(mean, median , mode, SD etc), probability, and probability distribution and sampling distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
data=pd.DataFrame({'values':[12,15,18,20,15,17,21,18,16,13,15,21]})
```

```python
mean_val=data['values'].mean()
median_val=data['values'].median()
mode_val=data['values'].mode()[0]
std_val=data['values'].std()

print('Mean:',mean_val,"Median:",median_val,"Mode:",mode_val,"SD:",std_val)
```

```python
prob=len(data[data['values']>15])/len(data)
print(prob)
```

```python
# Probability distribution
mean,std_dev=mean_val,std_val
normal=np.random.normal(mean, std_dev,1000)
sns.histplot(normal,bins=15,kde=True)
plt.title("Normal Distribution")
plt.show()
```

```python
# Sampling distribution
sample_mean=[]
for i in range(1000):
  sample=data['values'].sample(5,replace=True)
  sample_mean.append(sample.mean())

sns.histplot(sample_mean,bins=15,kde=True)
plt.title("Sampling Distribution")
plt.show()
```
